# Complexity

Repeated blur and sharpen operations do not cancel out. Small errors accumulate into visible image drift.

```python
from PIL import ImageFilter
from diffusers.utils import export_to_video, load_image

# Edit these inputs.
image_url = "https://picsum.photos/seed/64/512"
iterations = 100
output = "complexity.mp4"

image = load_image(image_url).convert("RGB")
frames = [image]

for _ in range(iterations):
    image = image.filter(ImageFilter.GaussianBlur(radius=2))
    image = image.filter(ImageFilter.UnsharpMask(radius=4, percent=150))
    frames.append(image)

export_to_video(frames, output, fps=30)
display(frames[0], frames[-1])
print(output)
```
